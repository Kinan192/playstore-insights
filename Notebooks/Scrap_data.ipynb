```python
import time
import pandas as pd
from google_play_scraper import Sort, reviews
```

```python
def scrap_n_reviews(app_id:str,
                     n:int = 3500,
                     lang:str = 'id', 
                     country: str = 'id',
                    sort=Sort.NEWEST, 
                    batch_size: int = 200, 
                    sleep_milliseconds: int = 1000
                    ):
    all_reviews = []
    token = None

    while len(all_reviews) < n:
        result, token = reviews(
            app_id,
            lang=lang,
            country=country,
            sort=sort,
            count=min(batch_size, n - len(all_reviews)),
            continuation_token=token
        )

        if not result:
            break

        all_reviews.extend(result)
        print(f'Scraped {len(all_reviews)} reviews so far...')

        if token is None:
            break

        time.sleep(sleep_milliseconds / 1000)
    
    uniq = {}
    for r in all_reviews:
        uniq[r.get("reviewId")] = r
    return list(uniq.values())
```

```python
app_id = 'com.traveloka.android'
data = scrap_n_reviews(app_id, n=3500, lang='id', country='id')
print(f'Total review terkumpul : {len(data)}')
```

```python
df = pd.DataFrame(data)
df.head(10)
```

```python
import re
import csv

def clean_text_cell(text):
    if pd.isna(text):
        return ""
    x = str(text)
    x = x.replace("\r", " ").replace("\n", " ")   # hilangkan enter dalam teks
    x = re.sub(r"\s+", " ", x).strip()            # rapikan spasi berlebih
    return x

df['content'] = df['content'].apply(clean_text_cell)
df['replyContent'] = df['replyContent'].apply(clean_text_cell)
```

```python
df.to_csv('data_tiketcom_3600_reviews.csv', index=False, encoding='utf-8-sig', quoting=csv.QUOTE_ALL)
print("Selesai menyimpan data ke file CSV.")
```
